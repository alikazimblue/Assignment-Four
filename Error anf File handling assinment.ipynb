### Q4 
Ek file `numbers.txt` banayein aur us mein 1 se 5 tak numbers line-by-line write karein.

```python
with open("numbers.txt", "w", encoding="utf-8") as file:
    for number in range(1, 6):
        file.write(f"{number}\n")

print("Numbers 1 to 5 written to numbers.txt")
```

### Q5
`numbers.txt` file ko read karein aur tamam numbers ka sum (total) calculate karke print karein.

```python
with open("numbers.txt", "r", encoding="utf-8") as file:
    numbers = [int(line.strip()) for line in file]

total = sum(numbers)
print("Numbers:", numbers)
print("Sum:", total)
```

### Q6 
Ek Python program likhein jo kisi file mein total lines count karke bataye.

```python
file_name = "test.txt"

try:
    with open(file_name, "r", encoding="utf-8") as file:
        line_count = sum(1 for _ in file)
    print(f"{file_name} contains {line_count} lines.")
except FileNotFoundError:
    print(f"Error: {file_name} was not found. Run Q1 first to create it.")
```

### Q7 (Medium)
`with open()` statement ka istemal karte hue ek file `data.txt` read karein aur explain karein ki `with` ka kya faida hai.

```python

with open("data.txt", "w", encoding="utf-8") as file:
    file.write("This is sample data in data.txt.\n")

try:
    with open("data.txt", "r", encoding="utf-8") as file:
        content = file.read()
    print(content)
except FileNotFoundError:
    print("data.txt was not found.")
```

### Q8 (Medium)
Ek file `sample.txt` se read karke uski saari lines ko ek Python list mein convert karein (`readlines()` ka use karke).

```python

with open("sample.txt", "w", encoding="utf-8") as file:
    file.write("First line\nSecond line\nThird line\n")

with open("sample.txt", "r", encoding="utf-8") as file:
    lines = file.readlines()

print(lines)
```

### Q9 
Ek file `source.txt` ka data read karke `destination.txt` file mein copy karein.

```python

with open("source.txt", "w", encoding="utf-8") as source_file:
    source_file.write("This content will be copied to destination.txt.\n")

with open("source.txt", "r", encoding="utf-8") as source_file:
    data = source_file.read()
with open("destination.txt", "w", encoding="utf-8") as destination_file:
    destination_file.write(data)

print("Data copied from source.txt to destination.txt")
```

### Q17
Ek `try-except` block likhein jo Multiple Exceptions catch kar sake (`ZeroDivisionError` aur `ValueError`).

```python
try:
    value = input("Enter a number (0 causes division by zero): ")
    number = int(value)
    print("10 divided by your number is:", 10 / number)
except (ZeroDivisionError, ValueError) as error:
    print("Handled error:", error)
```

### Q18
`else` block ka istemal `try-except` ke sath karein jo sirf tab chale jab koi error na aaye.

```python
try:
    number = int(input("Enter a number: "))
except ValueError:
    print("Please enter a valid whole number.")
else:
    print("No error occurred. Your number is", number)
```

### Q19 
`raise` keyword ka use karke age validation ka function banayein. Agar age 18 se kam ho toh custom `ValueError("Age must be 18 or older")` raise karein.

```python
def validate_age(age):
    if age < 18:
        raise ValueError("Age must be 18 or older")
    return "Age accepted"

try:
    age = int(input("Enter your age: "))
    print(validate_age(age))
except ValueError as error:
    print(error)
```

### Q20 
File handling aur Exception handling ko combine karein: Ek file read karein, agar file na mile toh error catch karein, aur `finally` block mein file closure handle karein.

```python

with open("data.txt", "w", encoding="utf-8") as setup_file:
    setup_file.write("This is sample data for the file handling example.\n")

file = None
try:
    file = open("data.txt", "r", encoding="utf-8")
    print(file.read())
except FileNotFoundError:
    print("Error: data.txt was not found.")
finally:
    if file is not None:
        file.close()
        print("File closed.")
```
